### Options

B. Replace image verbalization with Azure OpenAI text embeddings, and keep the existing vectorizer.

C. Store extracted images in a knowledge store, and query the image path field.

# Core Concept 1: Image Verbalization

## What is Image Verbalization?

Image verbalization converts an image into natural-language text.

### Workflow

```text
Diagram
   ↓
Image Verbalization
   ↓
Text Description
```

### Example

Diagram:

```text
User → API Gateway → Product Service → Database
```

Generated Description:

```text
Architecture diagram showing a user accessing a product service through an API gateway connected to a database.
```

---

## Benefits

✅ Natural language search

✅ RAG grounding

✅ Diagram explanation

✅ LLM-friendly content

---

## Limitation

Image verbalization does NOT create visual embeddings.

Therefore:

```text
Upload Image
      ↓
Search Similar Images
```

❌ Not Supported

---

# Core Concept 2: Multimodal Embeddings

## What are Multimodal Embeddings?

Multimodal embeddings convert images and text into vectors.

### Workflow

```text
Image
 ↓
Embedding Model
 ↓
Vector
```

Example:

```text
Network Diagram
       ↓
[0.32, -1.44, 0.81, ...]
```

Now Azure AI Search can compare vectors.

---

## Benefits

✅ Image Search

✅ Image Similarity Search

✅ Multimodal Retrieval

✅ Visual Matching

---

# Core Concept 3: Azure Vision Multimodal Embeddings Skill

## Purpose

Generates image vectors during indexing.

### Indexing Flow

```text
PDF Manual
      ↓
Extract Images
      ↓
Vision Multimodal Embeddings Skill
      ↓
Image Vectors
      ↓
Store in Search Index
```

### Used During

✅ Index Time

---

# Core Concept 4: Azure Vision Vectorizer

## Purpose

Converts uploaded images into query vectors.

### Query Flow

```text
Uploaded Image
       ↓
Vision Vectorizer
       ↓
Query Vector
       ↓
Vector Search
```

### Used During

✅ Query Time

```text
Find relevant content
```

```text
Help the LLM explain content
```

Example:

```text
Diagram
 ↓
Image Verbalization
 ↓
Text Description
 ↓
LLM
```

✅ Image Verbalization
