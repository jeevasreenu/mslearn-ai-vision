### Options

A. Azure Translator Text Translation API

B. A large language model prompt over extracted document text

C. Azure Translator Asynchronous Batch Document Translation

D. Azure Translator Synchronous Document Translation

---

# Correct Answer

✅ C. Azure Translator Asynchronous Batch Document Translation

---

# Key Clues

```text
Several hundred documents
Every night
Blob Storage
Preserve formatting
Target storage container
```

✅ Batch Processing

# AI-103 Note: Sentiment Analysis Document-Level Classification

## Question

You are building an app that analyzes user feedback using **Azure Language in Foundry Tools**.

A document named **Test.docx** contains:

- One positive sentence
- Multiple neutral sentences

You analyze the document using **Sentiment Analysis**.

### Which sentiment label will be returned for the document?

### Options

A. Mixed

B. Negative

C. Neutral

D. Positive

---

# Correct Answer

✅ D. Positive

---

# Key Clue

The question is testing:

```text
Document-level sentiment
```

not:

```text
Sentence-level sentiment
```

Many candidates incorrectly assume:

```text
Mostly neutral sentences
      ↓
Neutral document
```

This is NOT how Azure Sentiment Analysis works.

---

# Core Concept

Azure Sentiment Analysis evaluates:

### Sentence Level

Each sentence gets:

```text
Positive
Neutral
Negative
Mixed
```

---

### Document Level

Azure then determines an overall sentiment label.

A document containing:

```text
Positive Sentences
+
Neutral Sentences
```

is generally classified as:

✅ Positive

provided there are no negative sentences outweighing the sentiment.

---

# Example

## Input Document

```text
The application works very well.      ← Positive

The report was submitted yesterday.   ← Neutral

We reviewed the documents.            ← Neutral

The system processed the request.     ← Neutral
```

---

## Sentence Analysis

| Sentence | Sentiment |
|-----------|-----------|
| The application works very well | Positive |
| The report was submitted yesterday | Neutral |
| We reviewed the documents | Neutral |
| The system processed the request | Neutral |

---

## Document Result

```text
Positive
```

✅ Correct answer

---

# Why D is Correct

The document contains:

```text
At least one positive sentence
```

and

```text
All remaining sentences are neutral.
```

There are:

```text
No negative sentences
```

Therefore the overall document sentiment becomes:

✅ Positive

---

# Why A is Wrong

## Mixed

Mixed sentiment is typically returned when:

```text
Positive sentiment
+
Negative sentiment
```

exist together.

Example:

```text
The product is excellent.
The support team was terrible.
```

Result:

```text
Mixed
```

---

### Current Scenario

```text
Positive
+
Neutral
```

NOT

```text
Positive
+
Negative
```

❌ Wrong

---

# Why B is Wrong

## Negative

Negative requires:

```text
Negative sentiment present
```

Example:

```text
The application crashes constantly.
```

The question states:

```text
One positive sentence
Multiple neutral sentences
```

No negative sentiment exists.

❌ Wrong

---

# Why C is Wrong

## Neutral

Many students pick this answer because:

```text
Most sentences are neutral.
```

However Azure sentiment analysis is not simply:

```text
Majority vote
```

The positive sentiment influences the overall document sentiment.

Result:

✅ Positive

❌ Not Neutral

---

# Sentiment Categories

Example:

```text
Excellent service.
```

---

## Neutral

Example:

```text
The meeting starts at 10 AM.
```

---

instead of:

❌ Counting Sentences

✅ D. Positive

In Azure Language Sentiment Analysis, a document containing one positive sentence and the remaining neutral sentences is classified as **Positive**, because the positive sentiment influences the overall document-level sentiment result.

✅ Ensure output is parseable JSON

### What should you configure?

A. A prompt that requests valid JSON and includes an example narrative summary

B. A prompt that requests named entities and returns entities, topics, and sourceText

C. A prompt that requests valid JSON and a structured output schema with the accepted fields

D. A prompt that requests the accepted fields followed by a short explanation

---

# Correct Answer

✅ C. A prompt that requests valid JSON and a structured output schema with the accepted fields

---

# Key Clue

Look for:

```text
Downstream workflow
Only JSON accepted
Extra fields rejected
Parseable output required
```

This indicates:

✅ Structured Outputs

✅ JSON Schema Enforcement

✅ Controlled LLM Responses

---

Modern Foundry/OpenAI models support:

## Prompt Instructions

Tell the model:

```text
Return JSON
```

---

## Structured Output Schema

Tell the model exactly:

```text
What fields are allowed
What types are allowed
What structure is required
```

This greatly reduces:

❌ Hallucinated fields

❌ Narrative responses

❌ Parsing failures

---

# Why C is Correct

The requirement is:

```text
Return ONLY:

customer
product
location
topics
```

A structured schema enforces this format.

---

## Example Schema

```json
{
  "customer": "string",
  "product": "string",
  "location": "string",
  "topics": ["string"]
}
```

---

```text
Narrative content
```

❌ Wrong

# AI-103 Note: Azure Content Understanding - Extract vs Classify vs Generate

## Question

You have an Azure Content Understanding analyzer based on the prebuilt image analyzer.

The analyzer processes product images containing:

- Visible label text
- Packaging condition
- Visual context

You must return:

### Required Fields

```text
ProductName
Condition
Description
```

Where:

- ProductName = label text exactly as shown
- Condition = one of: new, used, damaged
- Description = short visual description

### Which field-method mapping should you use?

### Options

A. ProductName: generate ; Condition: classify ; Description: extract

B. ProductName: extract ; Condition: classify ; Description: generate

C. ProductName: extract ; Condition: generate ; Description: classify

D. ProductName: classify ; Condition: extract ; Description: generate

---

# Correct Answer

✅ B. ProductName: extract ; Condition: classify ; Description: generate

---

# Key Clue

The question tests the three core Content Understanding field methods:

```text
Extract
Classify
Generate
```

The trick is understanding **when each method should be used**.

---

# Core Concept

## Extract

Use when:

```text
The value already exists
in the source content.
```

Examples:

- Product names
- Invoice numbers
- Serial numbers
- IDs
- OCR text

The analyzer simply captures the value.

---

## Classify

Use when:

```text
The value must be chosen
from predefined categories.
```

Examples:

```text
New
Used
Damaged
```

Only one valid category is selected.

---

## Generate

Use when:

```text
The value must be created
from understanding the content.
```

Examples:

```text
Summaries
Descriptions
Captions
Explanations
```

The model generates new text.

---

# Requirement Analysis

## ProductName

Requirement:

```text
Label text as shown
```

Example Image Label:

```text
Contoso Wireless Headset
```

The text already exists in the image.

You want the exact value.

### Method

✅ Extract

---

## Condition

Requirement:

✅ First stage

Cannot happen until content exists.

Requires extracted content first.
