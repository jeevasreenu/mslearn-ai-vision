```python
import os
import json

from dotenv import load_dotenv
from tavily import TavilyClient

# =========================================================
# Setup
# =========================================================

load_dotenv()

client = TavilyClient(
    api_key="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"
)

# =========================================================
# Query
# =========================================================

query = """
Who won the Super Bowl in 2024?

In what state is the winning team's headquarters located?

What is the GDP of that state?

Answer each question clearly.
"""

# =========================================================
# Agentic Search
# =========================================================

result = client.search(
    query=query,
    max_results=5,
    include_answer=True,
)

# =========================================================
# AI Generated Answer
# =========================================================

print("\n" + "=" * 70)
print("AI GENERATED ANSWER")
print("=" * 70)

print(result.get("answer"))

# =========================================================
# Search Results
# =========================================================

print("\n" + "=" * 70)
print("SEARCH RESULTS")
print("=" * 70)

for i, item in enumerate(result.get("results", []), start=1):

    print(f"\nResult {i}")

    print("Title:")
    print(item.get("title"))

    print("\nURL:")
    print(item.get("url"))

    print("\nContent:")
    print(item.get("content"))

    print("\n" + "-" * 70)

# =========================================================
# Full JSON Response
# =========================================================

print("\n" + "=" * 70)
print("RAW RESPONSE")
print("=" * 70)

print(
    json.dumps(
        result,
        indent=2,
        ensure_ascii=False
    )
)
```
