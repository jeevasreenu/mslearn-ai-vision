# Azure AI Search - Hybrid Search and Semantic Ranker

## Question

You have an Azure AI Search index that contains clinical guideline chunks. The index includes:

- Text fields
- Vector embeddings
- Semantic configuration

User questions can include:

- Exact drug names
- Natural-language clinical descriptions

You need to configure retrieval for a grounding pipeline. The solution must:

1. Retrieve relevant chunks for both keyword and conceptual matches.
2. Improve the relevance ordering of the combined results.

### Which retrieval method and result-ordering method should you use?

---

## Key Requirements Analysis

### Requirement 1: Exact Keyword Matching

Example queries:

- Aspirin
- Ibuprofen
- Warfarin

These queries require finding documents that contain the exact terms.

**Solution:** Full-Text (Keyword) Search

---

### Requirement 2: Conceptual Matching

Example query:

> Medication used to prevent blood clots after surgery

The document may contain:

> Aspirin is recommended for preventing blood clots.

Even though the word "aspirin" is not present in the query, the system should still find the document.

**Solution:** Vector Search

---

### Requirement 3: Support Both Keyword and Conceptual Matches

The question explicitly states:

> Retrieve relevant chunks for both keyword and conceptual matches.

Neither keyword search nor vector search alone satisfies this requirement.

**Solution:** Hybrid Search

Hybrid Search combines:

- Full-Text Search (keywords)
- Vector Search (semantic similarity)

---

### Requirement 4: Improve Relevance Ordering

After Hybrid Search retrieves results, they must be ranked intelligently.

**Solution:** Semantic Ranker

Semantic Ranker:

- Understands user intent
- Understands context
- Reorders search results based on meaning
- Improves relevance of the final result set

---

## Correct Answer

### Retrieval Method

✅ Hybrid Search (Keyword + Vector Search)

### Result Ordering Method

✅ Semantic Ranker

---

## Why Hybrid Search?

### Keyword Search Only

Advantages:

- Finds exact drug names.
- Fast and precise for exact terms.

Disadvantages:

- Cannot understand meaning.
- Misses conceptually related documents.

❌ Not sufficient

---

### Vector Search Only

Advantages:

- Understands semantic meaning.
- Finds conceptually related content.

Disadvantages:

- May weaken exact keyword matching.

❌ Not sufficient

---

### Hybrid Search

Advantages:

- Supports exact keyword matches.
- Supports conceptual matches.
- Combines strengths of both retrieval methods.

✅ Correct choice

---

## Why Semantic Ranker?

Semantic Ranker performs:

- Semantic reranking
- Intent understanding
- Context-aware ordering

Example:

### Query

```
drug used after surgery to prevent blood clots
```

### Initial Hybrid Results

1. Surgical Recovery Procedures
2. Aspirin Dosage Guidelines
3. Blood Clot Prevention Recommendations

### Semantic Ranker Reorders

1. Blood Clot Prevention Recommendations
2. Aspirin Dosage Guidelines
3. Surgical Recovery Procedures

Most relevant results move to the top.

✅ Correct choice

---

## Why Other Options Are Wrong?

### Field Sorting

Example:

- Sort by Date
- Sort by Title

Only sorts based on field values.

Does not understand relevance.

❌ Wrong

---

### Scoring Profile

Used to:

- Boost certain fields
- Increase or decrease ranking weights

Example:

```json
{
  "title": 5,
  "content": 1
}
```

Limitation:

- Does not understand semantics.
- Does not perform semantic reranking.

❌ Wrong

---

## Exam Tips

### If the question mentions:

| Requirement | Answer |
|------------|---------|
| Exact match | Full-Text Search |
| Meaning-based search | Vector Search |
| Exact + Meaning | Hybrid Search |
| Improve relevance ranking | Semantic Ranker |
| Boost fields | Scoring Profile |
| Sort by field values | Field Sorting |

---

## Memory Shortcut

### Retrieval

```text
Keyword Search
      +
Vector Search
      =
 Hybrid Search
```

### Ranking

```text
Hybrid Search
      +
Semantic Ranker
      =
Best Search Experience
```

---

## Final Exam Answer

✅ Retrieval Method: Hybrid Search

✅ Result Ordering Method: Semantic Ranker

### One-Line Summary

Hybrid Search retrieves documents using both keyword and vector matching, while Semantic Ranker intelligently reranks the combined results based on meaning and relevance.

# Foundry Multi-Agent CI/CD - Preventing Incompatible Agent Promotions

## Question

You have a Microsoft Foundry project for a multi-agent customer-support system.

The code and infrastructure definitions are stored in GitHub.

You plan to use GitHub Actions to promote agent versions from staging to production.

Agents are versioned independently, and incompatible message schemas have caused failed handoffs between agents in staging.

You need to block production promotion when a changed agent is incompatible with a dependent agent.

### Which pipeline control should you recommend?

### Options

A. Add a deployment graph and staging contract tests for dependent agents.

B. Add health-check jobs for each agent after production deployment.

C. Add release-scoped rollback for the full multi-agent system.

D. Add parallel deployment jobs for all agents in the same environment.

---

## Correct Answer

✅ A. Add a deployment graph and staging contract tests for dependent agents.

---

## Step 1: Understand the Problem

The key issue is:

```text
Agents communicate with each other.
```

Example:

```text
Customer Agent
        ↓
Order Agent
        ↓
Billing Agent
```

A change in one agent may break another agent.

---

### Example of the Failure

Version 1:

```json
{
  "customerId": "123",
  "orderId": "456"
}
```

Version 2:

```json
{
  "customerNumber": "123",
  "orderId": "456"
}
```

The developer changed:

```text
customerId
```

to

```text
customerNumber
```

The dependent agent still expects:

```text
customerId
```

Result:

❌ Handoff failure

❌ Runtime errors

❌ Production outage

---

## Step 2: What Must Be Prevented?

The question states:

```text
Block production promotion
```

Important!

The solution must stop the deployment **before reaching production**.

This means:

✅ Validate compatibility in staging.

❌ Detect problems after deployment.

---

## Why A is Correct?

### Deployment Graph

A deployment graph tracks dependencies between agents.

Example:

```text
Agent A
   ↓
Agent B
   ↓
Agent C
```

If Agent B changes:

```text
Test with Agent A
Test with Agent C
```

before promotion.

---

### Contract Testing

Contract tests verify the schema and interface between agents.

Example:

Agent A promises:

```json
{
  "customerId": "123"
}
```

Contract test validates:

✅ Agent B still accepts this structure.

If the schema changes unexpectedly:

❌ Contract test fails.

❌ Deployment blocked.

✅ Prevents incompatible versions from reaching production.

---

## Why This Matches the Requirement

Requirement:

```text
Block promotion
```

Solution:

```text
Run dependency validation
+
Run contract tests
+
Fail pipeline if incompatible
```

✅ Perfect match

---

## Why B is Wrong?

### Health Checks After Deployment

Health checks verify:

- Service availability
- Endpoint responses
- Agent health

Example:

```text
GET /health
```

Returns:

```text
200 OK
```

Problem:

The deployment has already reached production.

Question requirement:

```text
Prevent promotion
```

Health checks occur too late.

❌ Wrong

---

## Why C is Wrong?

### Release-Scoped Rollback

Rollback means:

```text
Deploy
↓
Fail
↓
Revert
```

This helps recover from issues.

However:

The bad deployment already reached production.

Question requires:

```text
Prevent incompatible versions from being promoted.
```

Rollback fixes after the fact.

❌ Wrong

---

## Why D is Wrong?

### Parallel Deployment Jobs

Purpose:

- Faster deployments
- Reduced deployment time

Example:

```text
Deploy Agent A
Deploy Agent B
Deploy Agent C
```

simultaneously.

This does NOT validate:

- API contracts
- Message schemas
- Agent compatibility

❌ Wrong

---

```json
{
  "orderId": "123",
  "refundReason": "damaged"
}
```

✅ Production protected.
