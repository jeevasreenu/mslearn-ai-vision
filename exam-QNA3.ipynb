# AI-103 Note: DALL-E 3 Image Generation Request Body

## Question

You are building a web app that generates images based on user prompts.

The application uses the **DALL-E 3 Azure OpenAI model**.

You need to ensure that HTTP requests successfully generate images.

### Which HTTP body property should you include?

### Options

A. API Version

B. Azure OpenAI Resource Name

C. Deployment ID

D. Prompt

---

# Correct Answer

✅ D. Prompt

---

# Key Clue

The question asks:

```text
Which property belongs in the HTTP request BODY?
```

Many exam questions try to confuse:

- URL parameters
- Header values
- Request body properties

For DALL-E image generation, the required body property is:

✅ prompt

---

# Core Concept

An image generation request needs:

```text
What image should be generated?
```

The model gets this information from:

```json
{
  "prompt": "A futuristic city at sunset"
}
```

The prompt describes the image to create.

---

# Example Request

```http
POST /openai/deployments/dall-e-3/images/generations
```

Request Body:

```json
{
  "prompt": "A futuristic city at sunset"
}
```

The model uses:

```text
"A futuristic city at sunset"
```

to generate the image.

---

# Why D is Correct

The prompt contains:

✅ User instructions

✅ Image description

✅ Generation request

Without a prompt:

```text
The model has nothing to generate.
```

Therefore:

✅ Required in the body

---

# Why A is Wrong

## API Version

Example:

```text
api-version=2024-02-01
```

API version is typically passed as:

```text
Query Parameter
```

or endpoint configuration.

It is NOT the image-generation payload.

❌ Not the body property being requested.

---

# Why B is Wrong

## Azure OpenAI Resource Name

Example:

```text
my-openai-resource
```

This identifies:

```text
Azure Resource
```

and is part of:

```text
Endpoint URL
```

Example:

```text
https://my-openai-resource.openai.azure.com
```

Not part of the generation request body.

❌ Wrong

---

# Why C is Wrong

## Deployment ID

Example:

```text
dall-e-3
gpt-4o
```

Deployment ID appears in:

```http
/openai/deployments/{deployment-id}/
```

It is used to route the request.

It is NOT the image description.

❌ Wrong

---

# Visual Request Breakdown

## URL

```text
https://resource-name.openai.azure.com
```

Contains:

✅ Resource Name

---

## Endpoint Path

```text
/deployments/dall-e-3/
```

✅ Deployment ID

---

## Query Parameter

```text
?api-version=...
```

Contains:

✅ API Version

---

## Request Body

```json
{
  "prompt": "Generate a beach at sunset"
}
```

Contains:

✅ Prompt

---

# Final Exam Answer

✅ D. The Prompt

### One-Line Summary

For DALL-E 3 image generation, the **prompt** is the required HTTP request body property because it provides the textual description of the image the model should generate.

# AI-103 Note: Improving Retrieval Relevance in Azure OpenAI on Your Data

## Question

Your company is integrating Azure OpenAI in Foundry Models with an internal knowledge base to provide intelligent query responses.

You need to ensure that the **most relevant documents are prioritized during data retrieval**.

### What should you do?

### Options

A. Enable hybrid search mode

B. Increase retrieved document count

C. Reduce chunk size

D. Set strictness to a higher value

---

# Correct Answer

✅ D. Set Strictness to a Higher Value

---

# Key Clue

Look for:

```text
Most relevant documents
Prioritized during retrieval
```

The question is asking about:

✅ Retrieval quality

✅ Relevance filtering

NOT

✅ Number of documents

✅ Chunking

✅ Search method

---

# Core Concept

In Azure OpenAI on Your Data, **Strictness** controls how aggressively the retrieval system filters search results.

Higher strictness means:

```text
Only highly relevant documents
are returned to the model.
```

Lower strictness means:

```text
More documents returned
including less relevant matches.
```

---

# Visual Example

## User Query

```text
How do I reset my employee VPN password?
```

### Retrieval Results

| Document | Relevance |
|-----------|-----------|
| VPN Password Reset Guide | 95% |
| Remote Access Policy | 90% |
| Laptop Troubleshooting | 45% |
| Benefits Enrollment Guide | 20% |

---

## Low Strictness

Returns:

```text
VPN Password Reset Guide
Remote Access Policy
Laptop Troubleshooting
Benefits Enrollment Guide
```

More noise.

❌ Reduced relevance.

---

## High Strictness

Returns:

```text
VPN Password Reset Guide
Remote Access Policy
```

Only the strongest matches.

✅ Better relevance.

---

# Why D is Correct

The requirement is:

```text
Prioritize the most relevant documents
```

Higher strictness:

✅ Filters weak matches

✅ Improves grounding quality

✅ Reduces retrieval noise

✅ Improves RAG accuracy

---

# Why A is Wrong

## Hybrid Search

Hybrid Search combines:

```text
Keyword Search
+
Vector Search
```

Benefits:

✅ Better retrieval coverage

✅ Better recall

However:

The question asks:

```text
How do you prioritize relevance?
```

Hybrid search changes retrieval strategy.

It does NOT directly prioritize only the most relevant documents.

❌ Not the best answer

---

# Why B is Wrong

## Increase Retrieved Document Count

Example:

```text
Top 5 documents
→
Top 20 documents
```

Results:

More data

BUT

More low-quality matches.

---

Example:

```text
Relevant Answer
+
15 irrelevant documents
```

This may actually reduce answer quality.

❌ Opposite of the requirement

---

# Why C is Wrong

## Reduce Chunk Size

Chunk size determines:

```text
How documents are split
```

Example:

```text
200 words
```

vs

```text
1000 words
```

Smaller chunks may help granularity.

However:

They do NOT directly prioritize relevance.

❌ Wrong focus

---
